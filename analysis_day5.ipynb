# Day 5 – Python for Data: Pandas Basics
## Exploratory Data Analysis (EDA) and Data Cleaning

**Dataset:** Titanic passenger dataset (public sample dataset)  
**Tools:** Python, Pandas, Jupyter Notebook

### Objective
Load a CSV into a Pandas DataFrame, inspect and summarize the data, filter and select records, handle missing values, group and aggregate data, and export a cleaned CSV file.

## 1. Import Pandas and load the dataset

This notebook uses the public Titanic CSV hosted in the Seaborn example-data repository. An internet connection is required for the first download. If you have been given a different CSV by your internship, replace `DATA_URL` with your local file path.

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(DATA_URL)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

## 2. Inspect the dataset

```python
# First five rows
display(df.head())

# Last five rows
display(df.tail())

# Column names and data types, plus non-null counts
df.info()

# Summary statistics for numeric columns
display(df.describe())

# Summary statistics for all columns
display(df.describe(include="all"))
```

## 3. Check categories and missing values

```python
# Count the values in selected categorical columns
for column in ["survived", "sex", "class", "embark_town"]:
    if column in df.columns:
        print(f"\\nValue counts for {column}:")
        display(df[column].value_counts(dropna=False))

# Count missing values in each column
missing_counts = df.isnull().sum().sort_values(ascending=False)
display(missing_counts[missing_counts > 0])
```

## 4. Select columns and filter rows

```python
# Select a few useful columns
selected_columns = df[["survived", "pclass", "sex", "age", "fare", "embark_town"]]
display(selected_columns.head())

# Passengers aged 18 or above
adults = df[df["age"] >= 18]
print("Adult passengers:", len(adults))
display(adults.head())

# Female passengers who survived
female_survivors = df[(df["sex"] == "female") & (df["survived"] == 1)]
print("Female survivors:", len(female_survivors))
display(female_survivors.head())
```

## 5. Handle missing values

```python
# Work on a copy so the original DataFrame remains unchanged
cleaned_df = df.copy()

# Fill missing age values with the median age
cleaned_df["age"] = cleaned_df["age"].fillna(cleaned_df["age"].median())

# Fill missing embark_town values with the most frequent category (mode)
if cleaned_df["embark_town"].isnull().any():
    most_common_town = cleaned_df["embark_town"].mode()[0]
    cleaned_df["embark_town"] = cleaned_df["embark_town"].fillna(most_common_town)

# Drop deck because it has a large number of missing values in this dataset
if "deck" in cleaned_df.columns:
    cleaned_df = cleaned_df.drop(columns=["deck"])

print("Missing values after cleaning:")
display(cleaned_df.isnull().sum().sort_values(ascending=False).head(10))
```

## 6. Group by a column and aggregate

```python
# Survival rate and passenger count by gender
gender_summary = cleaned_df.groupby("sex").agg(
    passenger_count=("survived", "count"),
    survivors=("survived", "sum"),
    survival_rate=("survived", "mean"),
    average_age=("age", "mean")
).reset_index()

display(gender_summary)

# Average fare and survival rate by passenger class
class_summary = cleaned_df.groupby("pclass").agg(
    passenger_count=("survived", "count"),
    average_fare=("fare", "mean"),
    survival_rate=("survived", "mean")
).reset_index()

display(class_summary)
```

## 7. Sort and export cleaned data

```python
# Sort by passenger class and age
cleaned_df = cleaned_df.sort_values(
    by=["pclass", "age"],
    ascending=[True, True]
)

# Export without the DataFrame index
output_file = "titanic_cleaned_day5.csv"
cleaned_df.to_csv(output_file, index=False)

print(f"Cleaned dataset exported to: {output_file}")
```

## 8. Key observations

```python
# Generate a few observations from the dataset
overall_survival_rate = cleaned_df["survived"].mean()
print(f"Overall survival rate: {overall_survival_rate:.1%}")

print("\\nSurvival rate by gender:")
display(cleaned_df.groupby("sex")["survived"].mean().round(3))

print("\\nSurvival rate by passenger class:")
display(cleaned_df.groupby("pclass")["survived"].mean().round(3))
```
