```python
import pandas as pd

# First DataFrame
employees = pd.DataFrame({
    'EmployeeID': [1, 2, 3, 4],
    'Name': ['Vasavi', 'Ravi', 'Priya', 'Arun']
})

# Second DataFrame
departments = pd.DataFrame({
    'EmployeeID': [2, 3, 4, 5],
    'Department': ['IT', 'HR', 'Finance', 'Sales']
})

print("Employees DataFrame:")
print(employees)

print("\nDepartments DataFrame:")
print(departments)

# INNER JOIN
inner = pd.merge(employees, departments,
                 on='EmployeeID', how='inner')
print("\nINNER JOIN:")
print(inner)

# LEFT JOIN
left = pd.merge(employees, departments,
                on='EmployeeID', how='left')
print("\nLEFT JOIN:")
print(left)

# RIGHT JOIN
right = pd.merge(employees, departments,
                 on='EmployeeID', how='right')
print("\nRIGHT JOIN:")
print(right)

# OUTER JOIN
outer = pd.merge(employees, departments,
                 on='EmployeeID', how='outer')
print("\nOUTER JOIN:")
print(outer)

# CROSS JOIN
cross = pd.merge(employees, departments, how='cross')
print("\nCROSS JOIN:")
print(cross)
```

```python
import pandas as pd

sales_data = pd.DataFrame({
    'Region': ['North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile',
                'Mobile', 'Laptop', 'Mobile', 'Laptop'],
    'Sales': [50000, 30000, 45000, 25000,
              20000, 40000, 35000, 55000]
})

print("Original Sales Data:")
print(sales_data)

# Create Pivot Table
pivot = pd.pivot_table(
    sales_data,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print("\nPivot Table:")
print(pivot)
```

```python
import pandas as pd

# Wide format data
data = pd.DataFrame({
    'Student': ['A', 'B', 'C'],
    'Maths': [85, 90, 78],
    'Science': [88, 92, 80],
    'English': [82, 87, 85]
})

print("Wide Format:")
print(data)

# Convert wide format into long format
long_data = pd.melt(
    data,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Marks'
)

print("\nLong Format:")
print(long_data)
```

```python
import pandas as pd

# Employee data
data = pd.DataFrame({
    'Name': ['Vasavi', 'Ravi', 'Priya', 'Arun', 'Sneha'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT'],
    'Salary': [30000, 25000, 45000, 35000, 50000],
    'Experience': [1, 2, 3, 2, 4]
})

print("Original Data:")
print(data)

# Method chaining
result = (
    data
    .query("Department == 'IT'")
    .loc[lambda df: df['Salary'] > 30000]
    .sort_values(by='Salary', ascending=False)
    .reset_index(drop=True)
)

print("\nFinal Result:")
print(result)
```
