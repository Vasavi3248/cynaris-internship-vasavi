```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()

# Convert into DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add flower species
df['species'] = [
    iris.target_names[i] for i in iris.target
]

# Display first 5 rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
df.describe()
```

```python
sns.set_theme(style="whitegrid", palette="Set2")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df.index, df['sepal length (cm)'],
         color='blue', label='Sepal Length')

plt.title('Sepal Length Across Iris Samples')
plt.xlabel('Sample Number')
plt.ylabel('Sepal Length (cm)')
plt.legend()

plt.annotate('First sample',
             xy=(0, df['sepal length (cm)'][0]),
             xytext=(15, 4.5),
             arrowprops=dict(arrowstyle='->'))

plt.savefig('line_chart.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='species',
            y='petal length (cm)')

plt.title('Average Petal Length by Species')
plt.xlabel('Flower Species')
plt.ylabel('Petal Length (cm)')

plt.savefig('bar_chart.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df,
                x='petal length (cm)',
                y='petal width (cm)',
                hue='species',
                s=70)

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.legend(title='Species')

plt.savefig('scatter_plot.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df,
             x='sepal length (cm)',
             bins=15,
             kde=True)

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')

plt.savefig('histogram.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df,
            x='species',
            y='petal length (cm)')

plt.title('Petal Length Distribution by Species')
plt.xlabel('Flower Species')
plt.ylabel('Petal Length (cm)')

plt.savefig('box_plot.png', dpi=300)
plt.show()
```

```python
## Data Visualisation – Observations

1. **Line Chart:** Displays changes in sepal length across 150 Iris samples.
2. **Bar Chart:** Compares average petal length among three flower species.
3. **Scatter Plot:** Shows the relationship between petal length and petal width. Different species form distinct clusters.
4. **Histogram:** Shows the distribution and frequency of sepal lengths.
5. **Box Plot:** Compares petal length distributions and identifies potential outliers among flower species.

### Conclusion
Matplotlib and Seaborn were used to create five different data visualisations using the Iris dataset. These charts help identify patterns, distributions, comparisons, and relationships in the data.
```
